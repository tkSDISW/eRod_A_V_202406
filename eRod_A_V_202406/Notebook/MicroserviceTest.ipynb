```python
from flask import Flask, request, jsonify
import threading

# Create Flask app
app = Flask(__name__)

# Define an endpoint for processing data
@app.route('/process', methods=['POST'])
def process_data():
    # Extract input from request JSON
    data = request.json
    number = data.get('number')
    if number is None:
        return jsonify({'error': 'No number provided'}), 400

    # Perform processing (e.g., square the number)
    result = number ** 2
    return jsonify({'number': number, 'result': result})

# Function to run Flask app in a separate thread
def run_app():
    app.run(port=5000, debug=False, use_reloader=False)

# Start Flask app in a background thread
flask_thread = threading.Thread(target=run_app)
flask_thread.daemon = True
flask_thread.start()

print("Microservice is running on http://127.0.0.1:5000")
```

```python
import requests

# Define the API endpoint
url = "http://127.0.0.1:5000/process"

# Test data
data = {"number": 5}

# Send a POST request to the microservice
response = requests.post(url, json=data)

# Print the response
if response.status_code == 200:
    print("Response:", response.json())
else:
    print("Error:", response.status_code, response.text)
```
