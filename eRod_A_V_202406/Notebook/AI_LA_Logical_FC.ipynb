```python
import numpy as np
import matplotlib.pyplot as plt

# Define the functions in the braking system

def apply_brake(pedal_force):
    """Converts driver's pedal force to braking force."""
    return pedal_force * 0.8  # Conversion efficiency

def brake_force_to_hydraulic_force(brake_force):
    """Converts brake force to hydraulic force in the brake master cylinder."""
    return brake_force * 0.9  # Conversion efficiency

def amplify_hydraulic_force(hydraulic_force):
    """Amplifies the hydraulic force for distribution."""
    return hydraulic_force * 1.5

def distribute_force(hydraulic_force):
    """Distributes the force equally to all wheels."""
    return hydraulic_force / 4

def generate_traction(wheel_force):
    """Generates traction at each wheel."""
    terrain_friction = 0.8  # Coefficient of friction of the terrain
    return wheel_force * terrain_friction

# Simulate the braking system

def simulate_braking_system(pedal_force):
    brake_force = apply_brake(pedal_force)
    hydraulic_force = brake_force_to_hydraulic_force(brake_force)
    amplified_force = amplify_hydraulic_force(hydraulic_force)
    wheel_force = distribute_force(amplified_force)
    traction = generate_traction(wheel_force)
    
    return {
        "Pedal Force (N)": pedal_force,
        "Brake Force (N)": brake_force,
        "Hydraulic Force (N)": hydraulic_force,
        "Amplified Force (N)": amplified_force,
        "Wheel Force per Wheel (N)": wheel_force,
        "Traction per Wheel (N)": traction
    }

# Test the simulation with varying pedal forces
pedal_forces = np.linspace(0, 500, 100)  # Range of pedal forces from 0 to 500 N
results = [simulate_braking_system(pf) for pf in pedal_forces]

# Extract data for plotting
brake_forces = [result["Brake Force (N)"] for result in results]
hydraulic_forces = [result["Hydraulic Force (N)"] for result in results]
amplified_forces = [result["Amplified Force (N)"] for result in results]
wheel_forces = [result["Wheel Force per Wheel (N)"] for result in results]
tractions = [result["Traction per Wheel (N)"] for result in results]

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(pedal_forces, brake_forces, label="Brake Force (N)")
plt.plot(pedal_forces, hydraulic_forces, label="Hydraulic Force (N)")
plt.plot(pedal_forces, amplified_forces, label="Amplified Force (N)")
plt.plot(pedal_forces, wheel_forces, label="Wheel Force per Wheel (N)")
plt.plot(pedal_forces, tractions, label="Traction per Wheel (N)")
plt.xlabel("Pedal Force (N)")
plt.ylabel("Force (N)")
plt.title("Manual Braking System Simulation")
plt.legend()
plt.grid()
plt.show()
```

Change the simulation so that class are used for all the components and actors and functions are methods of the components.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define classes for each component in the braking system

class Driver:
    def __init__(self, pedal_force):
        self.pedal_force = pedal_force

    def apply_brake(self):
        """Converts driver's pedal force to braking force."""
        return self.pedal_force * 0.8  # Conversion efficiency

class BrakeMasterCylinder:
    def __init__(self):
        self.brake_force = 0

    def brake_force_to_hydraulic_force(self, brake_force):
        """Converts brake force to hydraulic force."""
        self.brake_force = brake_force
        return self.brake_force * 0.9  # Conversion efficiency

class HydraulicControlModulator:
    def __init__(self):
        self.hydraulic_force = 0

    def amplify_hydraulic_force(self, hydraulic_force):
        """Amplifies the hydraulic force for distribution."""
        self.hydraulic_force = hydraulic_force
        return self.hydraulic_force * 1.5

class BrakeLinesAndFluid:
    def __init__(self):
        self.amplified_force = 0

    def distribute_force(self, amplified_force):
        """Distributes the force equally to all wheels."""
        self.amplified_force = amplified_force
        return self.amplified_force / 4

class Wheel:
    def __init__(self):
        self.wheel_force = 0

    def generate_traction(self, wheel_force):
        """Generates traction at each wheel."""
        self.wheel_force = wheel_force
        terrain_friction = 0.8  # Coefficient of friction of the terrain
        return self.wheel_force * terrain_friction

# Simulate the braking system

def simulate_braking_system(pedal_force):
    driver = Driver(pedal_force)
    brake_master_cylinder = BrakeMasterCylinder()
    hydraulic_modulator = HydraulicControlModulator()
    brake_lines = BrakeLinesAndFluid()
    wheel = Wheel()

    brake_force = driver.apply_brake()
    hydraulic_force = brake_master_cylinder.brake_force_to_hydraulic_force(brake_force)
    amplified_force = hydraulic_modulator.amplify_hydraulic_force(hydraulic_force)
    wheel_force = brake_lines.distribute_force(amplified_force)
    traction = wheel.generate_traction(wheel_force)

    return {
        "Pedal Force (N)": pedal_force,
        "Brake Force (N)": brake_force,
        "Hydraulic Force (N)": hydraulic_force,
        "Amplified Force (N)": amplified_force,
        "Wheel Force per Wheel (N)": wheel_force,
        "Traction per Wheel (N)": traction
    }

# Test the simulation with varying pedal forces
pedal_forces = np.linspace(0, 500, 100)  # Range of pedal forces from 0 to 500 N
results = [simulate_braking_system(pf) for pf in pedal_forces]

# Extract data for plotting
brake_forces = [result["Brake Force (N)"] for result in results]
hydraulic_forces = [result["Hydraulic Force (N)"] for result in results]
amplified_forces = [result["Amplified Force (N)"] for result in results]
wheel_forces = [result["Wheel Force per Wheel (N)"] for result in results]
tractions = [result["Traction per Wheel (N)"] for result in results]

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(pedal_forces, brake_forces, label="Brake Force (N)")
plt.plot(pedal_forces, hydraulic_forces, label="Hydraulic Force (N)")
plt.plot(pedal_forces, amplified_forces, label="Amplified Force (N)")
plt.plot(pedal_forces, wheel_forces, label="Wheel Force per Wheel (N)")
plt.plot(pedal_forces, tractions, label="Traction per Wheel (N)")
plt.xlabel("Pedal Force (N)")
plt.ylabel("Force (N)")
plt.title("Manual Braking System Simulation")
plt.legend()
plt.grid()
plt.show()
```
