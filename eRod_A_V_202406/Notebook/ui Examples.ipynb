```python
import ipywidgets as widgets
from IPython.display import display

# Create a slider to simulate the brake pedal
slider = widgets.FloatSlider(
    value=0,
    min=0,
    max=100,
    step=1,
    description='Brake Pedal:',
)

# Create an LED indicator
led = widgets.Label(
    value="🔴",  # Red circle emoji for OFF state
    layout=widgets.Layout(width='50px')
)

# Function to update the LED based on slider value
def update_led(change):
    if change['new'] > 50:
        led.value = "🟢"  # Green circle emoji for ON state
    else:
        led.value = "🔴"  # Red circle emoji for OFF state

# Observe changes in slider value
slider.observe(update_led, names='value')

# Display widgets
display(slider, led)
```

```python
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, interact
import numpy as np

# Function to update the dial visualization
def update_dial(value):
    plt.clf()
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.set_xlim(-1, 1)
    ax.set_ylim(-1, 1)
    ax.axis('off')

    # Draw the outer dial
    circle = plt.Circle((0, 0), 0.9, color='lightgray')
    ax.add_artist(circle)

    # Draw the needle
    angle = np.deg2rad(value * 1.8 - 90)  # Map 0-100 to -90 to +90 degrees
    needle_x = [0, 0.8 * np.cos(angle)]
    needle_y = [0, 0.8 * np.sin(angle)]
    ax.plot(needle_x, needle_y, color='red', linewidth=3)

    # Add tick marks
    for tick_angle in range(-90, 91, 30):
        tick_rad = np.deg2rad(tick_angle)
        x1, y1 = 0.85 * np.cos(tick_rad), 0.85 * np.sin(tick_rad)
        x2, y2 = 0.95 * np.cos(tick_rad), 0.95 * np.sin(tick_rad)
        ax.plot([x1, x2], [y1, y2], color='black')

    # Add text annotation
    ax.text(0, -1.2, f"Value: {value:.1f}", ha='center', fontsize=12)

    plt.title("Interactive Dial")
    plt.show()

# Create a slider widget
slider = FloatSlider(value=50, min=0, max=100, step=1, description="Dial:")

# Bind the slider to the dial visualization
interact(update_dial, value=slider)
```

```python
# Create buttons
buttons = [
    widgets.ToggleButton(
        value=False,
        description=f"Button {i}",
        button_style="",  # 'success' for green buttons
        tooltip=f"Button {i}",
    )
    for i in range(1, 5)
]

# Function to handle button selection
def on_button_click(change):
    # Ensure only one button can be active
    for btn in buttons:
        if btn != change["owner"]:
            btn.value = False
    # Update button styles
    for btn in buttons:
        btn.button_style = "success" if btn.value else ""

# Attach handlers to buttons
for button in buttons:
    button.observe(on_button_click, names="value")

# Display the buttons in a horizontal layout
display(widgets.HBox(buttons))
```

```python
led = widgets.Label(
    value="🔴",  # Red circle emoji for OFF state
    layout=widgets.Layout(width="50px")
)

# Create buttons
button_on = widgets.Button(description="Turn ON", button_style="success")
button_off = widgets.Button(description="Turn OFF", button_style="danger")

# Define button callbacks
def turn_on_led(b):
    led.value = "🟢"  # Green circle emoji for ON state

def turn_off_led(b):
    led.value = "🔴"  # Red circle emoji for OFF state

# Assign callbacks to buttons
button_on.on_click(turn_on_led)
button_off.on_click(turn_off_led)

# Display widgets
display(led)
display(widgets.HBox([button_on, button_off]))
```
